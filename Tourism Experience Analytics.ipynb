# 📊 Dataset Explanation

## Tourism Dataset

The **Tourism Dataset** is a structured collection of tourism transaction, user, geographical, and attraction information. It combines user profiles with historical attraction visits and ratings, providing the foundation for **tourism analytics, visit-mode classification, attraction-rating prediction, and personalized recommendations**.

## 1. 🧾 Transaction Data

**Purpose:**
Contains information about users' visits to tourist attractions, including visit timing, visit mode, attraction, and rating.

**Key Columns:**

* `TransactionId` — Unique transaction identifier
* `UserId` — Identifies the user
* `VisitYear` — Year of the visit
* `VisitMonth` — Month of the visit
* `VisitMode` — Type of visit, such as Business, Family, Couples, etc.
* `AttractionId` — Identifies the visited attraction
* `Rating` — User's rating for the attraction

**Business Use:**
Used to understand visitor behavior, predict visit modes, analyze satisfaction, and generate attraction recommendations.

---

## 2. 👤 User Data

**Purpose:**
Provides geographical and demographic information about users.

**Key Columns:**

* `UserId`
* `ContinentId`
* `RegionId`
* `CountryId`
* `CityId`

**Business Use:**
Helps identify geographical patterns in tourism behavior and provides features for visit-mode prediction and recommendation.

---

## 3. 🏙️ City Data

**Key Columns:**

## 🧠 7. ML Preprocessing

We split the data **before** computing any target-derived statistics (attraction/user average rating) to avoid leaking test-set information into training features.


````python
cat_cols = ["Continent", "Region", "Country", "AttractionCategory"]
encoders = {}
for c in cat_cols:
    le = LabelEncoder()
    master[c + "_enc"] = le.fit_transform(master[c].astype(str))
    encoders[c] = le

target_le = LabelEncoder()
master["VisitModeTarget"] = target_le.fit_transform(master["VisitModeLabel"])

train_df, test_df = train_test_split(master, test_size=0.2, random_state=RANDOM_STATE, stratify=master["VisitModeTarget"])

# Leakage-safe aggregate features: computed on TRAIN ONLY, then mapped onto both splits
attr_avg_train = train_df.groupby("AttractionId")["Rating"].mean()
user_avg_train = train_df.groupby("UserId")["Rating"].mean()
global_avg = train_df["Rating"].mean()

for d in (train_df, test_df):
    d["AttractionAvgRating"] = d["AttractionId"].map(attr_avg_train).fillna(global_avg)
    d["UserAvgRating"] = d["UserId"].map(user_avg_train).fillna(global_avg)

print("Train:", train_df.shape, " Test:", test_df.shape)
````

## 📈 8. Regression — Predicting Attraction Ratings


````python
reg_features = ["VisitYear", "VisitMonth", "Continent_enc", "Region_enc",
                "Country_enc", "AttractionCategory_enc", "AttractionAvgRating"]

X_train_r, y_train_r = train_df[reg_features], train_df["Rating"]
X_test_r,  y_test_r  = test_df[reg_features],  test_df["Rating"]

reg_models = {
    "Linear Regression": LinearRegression(),
    "Random Forest":     RandomForestRegressor(n_estimators=200, max_depth=10, random_state=RANDOM_STATE, n_jobs=-1),
    "XGBoost":           xgb.XGBRegressor(n_estimators=250, max_depth=6, learning_rate=0.08, random_state=RANDOM_STATE),
    "LightGBM":          lgb.LGBMRegressor(n_estimators=250, random_state=RANDOM_STATE, verbose=-1),
}

reg_results = []
fitted_reg_models = {}
for name, model in reg_models.items():
    model.fit(X_train_r, y_train_r)
    preds = model.predict(X_test_r)
    reg_results.append({
        "Model": name,
        "R2": r2_score(y_test_r, preds),
        "RMSE": mean_squared_error(y_test_r, preds) ** 0.5,
        "MAE": mean_absolute_error(y_test_r, preds),
    })
    fitted_reg_models[name] = model

reg_results_df = pd.DataFrame(reg_results).sort_values("R2", ascending=False)
reg_results_df
````

## 🤖 10. Recommendation Engine

Only **30 attractions** have transaction history, so a 33.5K × 33.5K user-user similarity matrix would be both wasteful and memory-prohibitive. Instead we build a lightweight, production-realistic **item-based collaborative filter** (30 × 30 similarity — trivial to compute and refresh) plus a **content-based fallback** for cold-start attractions.


````python
user_item = train_df.pivot_table(index="UserId", columns="AttractionId", values="Rating", aggfunc="mean").fillna(0)
print("User-item matrix:", user_item.shape)

item_sim = cosine_similarity(user_item.T.values)
item_sim_df = pd.DataFrame(item_sim, index=user_item.columns, columns=user_item.columns)

attraction_names = master.drop_duplicates("AttractionId").set_index("AttractionId")["Attraction"]
````

````python
def recommend_collaborative(user_id, n=5):
    """Item-based CF: score unseen attractions by similarity to what the user already rated highly."""
    if user_id not in user_item.index:
        return pd.Series(dtype=float)
    rated = user_item.loc[user_id]
    rated = rated[rated > 0]
    if rated.empty:
        return pd.Series(dtype=float)
    scores = pd.Series(0.0, index=user_item.columns)
    for attraction_id, rating in rated.items():
        scores += item_sim_df[attraction_id] * rating
    scores = scores.drop(index=rated.index)
    return scores.sort_values(ascending=False).head(n)

sample_user = user_item.index[0]
top = recommend_collaborative(sample_user)
result = pd.DataFrame({
    "Attraction": [attraction_names.get(i, i) for i in top.index],
    "Score": top.values
})
print(f"Top recommendations for User {sample_user}:")
result
````

# 🔄 Content-based fallback — useful for attractions with little/no rating history

````python
attraction_features = (
    master.drop_duplicates("AttractionId")[["AttractionId", "AttractionCategory_enc"]]
    .merge(master.groupby("AttractionId")["Rating"].mean().rename("AvgRating"), on="AttractionId")
    .set_index("AttractionId")
)
content_sim = cosine_similarity(attraction_features.values)
content_sim_df = pd.DataFrame(content_sim, index=attraction_features.index, columns=attraction_features.index)

def recommend_content_based(attraction_id, n=5):
    if attraction_id not in content_sim_df.index:
        return pd.Series(dtype=float)
    sims = content_sim_df[attraction_id].drop(index=attraction_id)
    return sims.sort_values(ascending=False).head(n)

sample_attraction_id = attraction_features.index[0]
top_c = recommend_content_based(sample_attraction_id)
result_c = pd.DataFrame({
    "Similar Attraction": [attraction_names.get(i, i) for i in top_c.index],
    "Similarity": top_c.values
})
print(f"Attractions similar to '{attraction_names.get(sample_attraction_id)}':")
result_c
````

````python
fig = px.imshow(
    item_sim_df.rename(index=attraction_names, columns=attraction_names),
    color_continuous_scale="Viridis", aspect="auto",
    title="Item-Item Similarity Matrix (Collaborative Filtering Backbone)"
)
fig.update_layout(height=650)
fig.show()
````

**Approach recap:**
- **Collaborative filtering** captures *behavioral* similarity ("people who liked X also liked Y") — best when an attraction has rating history.
- **Content-based filtering** captures *attribute* similarity (same category, similar average rating) — solves cold-start for new/unrated attractions.
- A production system would blend both (hybrid), weighting collaborative signal higher as an attraction accumulates ratings.
